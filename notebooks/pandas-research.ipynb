````python
# функция для подсчета времени работы

def count_time(func, *data):
    tick = time.perf_counter()
    func(*data)
    return time.perf_counter() - tick
````

````python
# получим время работы функций

# max_prod_mod_3

max_prod_mod_3_time = []
max_prod_mod_3_vectorised_time = []

for x in max_prod_mod_3_data:
  max_prod_mod_3_time += [count_time(research_functions.max_prod_mod_3, x.tolist())]
  max_prod_mod_3_vectorised_time += [count_time(research_functions_vectorised.max_prod_mod_3, x)]

sizes_max_prod_mod_3 = [10, 100, 1000, 10000]

# are_multisets_equal
are_multisets_equal_time = []
are_multisets_equal_vectorised_time = []

for x, y in are_multisets_equal_data:
    are_multisets_equal_time += [count_time(research_functions.are_multisets_equal, x.tolist(), y.tolist())]
    are_multisets_equal_vectorised_time += [count_time(research_functions_vectorised.are_multisets_equal, x, y)]

sizes_are_multisets_equal = [10, 100, 1000, 10000]

# convert_image
convert_image_time = []
convert_image_vectorised_time = []

for image, weights in convert_image_data:
    convert_image_time += [count_time(research_functions.convert_image, image.tolist(), weights.tolist())]
    convert_image_vectorised_time += [count_time(research_functions_vectorised.convert_image, image, weights)]

sizes_convert_image = [10, 100, 1000]

# rle_scalar
rle_scalar_time = []
rle_scalar_vectorised_time = []

for x, y in rle_scalar_data:
    rle_scalar_time += [count_time(research_functions.rle_scalar, x.tolist(), y.tolist())]
    rle_scalar_vectorised_time += [count_time(research_functions_vectorised.rle_scalar, x, y)]

sizes_rle_scalar = [10, 100, 1000, 10000]

# cosine_distance
cosine_distance_time = []
cosine_distance_vectorised_time = []

for X, Y in cosine_distance_data:
    cosine_distance_time += [count_time(research_functions.cosine_distance, X.tolist(), Y.tolist())]
    cosine_distance_vectorised_time += [count_time(research_functions_vectorised.cosine_distance, X, Y)]

sizes_cosine_distance = [10, 100]
````

````python
# ВАШИ ГРАФИКИ ЗДЕСЬ
import matplotlib.pyplot as plt

def plot_in_subplot(ax, sizes, non_vectorised_time, vectorised_time, title):
    ax.plot(sizes, non_vectorised_time, label='Невекторизованная реализация', marker='o', color='red')
    ax.plot(sizes, vectorised_time, label='Векторизованная реализация', marker='o', color='green')
    ax.set_xlabel('Объем данных')
    ax.set_ylabel('Время выполнения (с)')
    ax.set_xscale('log')
    ax.set_title(title)
    ax.legend()
    ax.grid(True)

fig, axs = plt.subplots(3, 2, figsize=(10, 10))

plot_in_subplot(axs[0, 0], sizes_max_prod_mod_3, max_prod_mod_3_time, max_prod_mod_3_vectorised_time, 'Max Prod Mod 3')
plot_in_subplot(axs[0, 1], sizes_are_multisets_equal, are_multisets_equal_time, are_multisets_equal_vectorised_time, 'Are Multisets Equal')
plot_in_subplot(axs[1, 0], sizes_convert_image, convert_image_time, convert_image_vectorised_time, 'Convert Image')
plot_in_subplot(axs[1, 1], sizes_rle_scalar, rle_scalar_time, rle_scalar_vectorised_time, 'RLE Scalar')
plot_in_subplot(axs[2, 0], sizes_cosine_distance, cosine_distance_time, cosine_distance_vectorised_time, 'Cosine Distance')

fig.delaxes(axs[2, 1])
plt.tight_layout()
````
