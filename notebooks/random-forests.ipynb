# Случайные леса

Как многие из вас уже знают, теорема Кондорсе о присяжных ,
гласит, что если у нас имеется некоторое количество различных методов,
каждый из которых предсказывает величину лучше, чем случайно, то
голосование таких методов работает лучше, чем любой из этих методов по
отдельности.

Данная идея отлично ложится на деревья решений. В теории для
повышения дисперсии предсказаний каждого отдельного дерева можно
ограничивать их глубину. На практике, помимо этого, также хорошо себя
показало использование ограниченного набора признаков и данных. Это
приводит нас к идее случайных лесов, о которых вы узнали на занятии.

А что влияет на улучшение, которое мы получаем при использовании
ансамбля методов? Соответствующая теорема дает достаточно простой ответ:

предсказания которых отличаются незначительно, то их усреднение
практически ничего не изменит. Если же у нас есть множество достаточно
разных методов, то и их усреднение будет отличаться от предсказаний
каждого отдельного метода. Видимо, random_state было недостаточно. Как
же нам еще сильнее увеличить различность методов? Ответ вы знаете:
использовать случайные признаки и случайную выборку объектов. Это

время, потраченное на обучение. Для обучаемых лесов, используйте число

4. Постройте еще один график для отображения зависимости значения MAE
от числа деревьев в лесу. Вдоль оси Ox отложите число деревьев,

частности - переобучение)

````python
import math

forests_time = []
forests_mae_train = []
forests_mae_test = []
for i in np.logspace(1, 3, num=10):
  regr = RandomForestRegressor(n_estimators=math.floor(i), random_state=42)
  time_1 = time.time()
  regr.fit(X_train, y_train)
  time_2 = time.time()
  forests_time.append(time_2 - time_1)

  predict = regr.predict(X_train)
  forests_mae_train.append(mean_absolute_error(predict, y_train))

  predict = regr.predict(X_test)
  forests_mae_test.append(mean_absolute_error(predict, y_test))
````

````python
num_of_trees = [math.floor(i) for i in np.logspace(1, 3, num=10)]
plt.plot(num_of_trees, forests_time)
plt.xlabel('Число деревьев')
plt.ylabel('Время (мс)')
plt.title('Зависимость времени обучения от числа деревьев в лесу')
plt.show()
````

````python
plt.plot(num_of_trees, forests_mae_train, label='train')
plt.plot(num_of_trees, forests_mae_test, label='test')
plt.xlabel('Число деревьев')
plt.ylabel('MAE')
plt.legend()
plt.grid()
plt.title('Зависимость MAE от числа деревьев в лесу')
plt.show()
````

4. Выведите признаки, которые были занулены. Имена признаков можно найти в fetch_california_housing().feature_names
5. Сделайте выводы о способности LASSO отбирать признаки

````python
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler
````

````python
lin = [Pipeline([
    ('scaler', StandardScaler()),
    ('regressor', Lasso(alpha=i))
]).fit(X_train, y_train) for i in [0.1, 0.2, 0.5]]

print('alpha=0.1')
print('MAE =', mean_absolute_error(y_test, lin[0].predict(X_test)))
zeros = np.where(lin[0].named_steps['regressor'].coef_ == 0)[0]
print('num of non zero ', len(lin[0].named_steps['regressor'].coef_) - zeros.size)
print('zero in ', [np.array(fetch_california_housing().feature_names)[i] for i in zeros])

print('alpha=0.2')
print('MAE =', mean_absolute_error(y_test, lin[1].predict(X_test)))
zeros = np.where(lin[1].named_steps['regressor'].coef_== 0)[0]
print('num of non zero ', len(lin[1].named_steps['regressor'].coef_) - zeros.size)
print('zero in ', [np.array(fetch_california_housing().feature_names)[i] for i in zeros])

print('alpha=0.5')
print('MAE =', mean_absolute_error(y_test, lin[2].predict(X_test)))
zeros = np.where(lin[2].named_steps['regressor'].coef_== 0)[0]
print('num of non zero ', len(lin[2].named_steps['regressor'].coef_) - zeros.size)
print('zero in ', [np.array(fetch_california_housing().feature_names)[i] for i in zeros])
````

Вывод по результату :

Lasso отбирает признаки, при этом на жесткость отбора влияет

признаков.

Отлично! Мы убедились, что при помощи L1 регуляризации можно отбирать
наиболее информативные признаки. Такой же трюк можно применять к svm,
который может занулить признак, если ни один опорный объект не
использует его.

Как мы помним, при помощи линейных классификаторов можно ранжировать
признаки по важности, воспользовавшись соответствующими коэффициентами.
Но перед этим нужно нормализовать используемые признаки, иначе значения

### **Задание 2.3 [2 балл]**

Данные : датасет со стоимостью недвижимости в Калифорнии
Метрика качества : MAE
Цели : В данном задании следует выполнить следующие пункты:

1. Реализуйте процесс отбора признаков путем добавление одного наиболее информативного признака. Воспользуйтесь DecisionTreeRegressor в качестве базового классификатора.
2. Отберите 3 наиболее важных признака
3. Обучите обычную линейную регрессию на всех признаках, а также только
на отобранных признаках. Сравните качество предсказаний и сделайте
вывод о качестве выделенных признаков. Сравните с признаками,
выделенными в предыдущем задании.

````python
def select_best_features(X_train, X_test, y_train, y_test, max_features=3):
    best_features = []
    remaining_features = list(range(X_train.shape[1]))
    feature_names= fetch_california_housing().feature_names
    for _ in range(max_features):
        best_feature = None
        best_mae = np.inf

        for feature in remaining_features:
            current_features = best_features + [feature]

            model = DecisionTreeRegressor().fit(X_train[:, current_features], y_train)
            y_pred = model.predict(X_test[:, current_features])
            mae = mean_absolute_error(y_test, y_pred)

            if mae < best_mae:
                best_mae = mae
                best_feature = feature

        best_features.append(best_feature)
        remaining_features.remove(best_feature)
        print(f"Добавлен признак: {feature_names[best_feature]}, MAE: {best_mae:.4f}")

    return best_features
````

````python
best_feature_indices = select_best_features(
    X_train, X_test, y_train, y_test, max_features=3)

best_features_names = [feature_names[idx] for idx in best_feature_indices]
print("\nОтобранные признаки:", best_features_names)

scaler_all = StandardScaler()
X_train_scaled_all = scaler_all.fit_transform(X_train)
X_test_scaled_all = scaler_all.transform(X_test)

scaler_selected = StandardScaler()
X_train_scaled_selected = scaler_selected.fit_transform(X_train[:, best_feature_indices])
X_test_scaled_selected = scaler_selected.transform(X_test[:, best_feature_indices])

lr_all = LinearRegression()
lr_all.fit(X_train_scaled_all, y_train)
mae_all = mean_absolute_error(y_test, lr_all.predict(X_test_scaled_all))

lr_selected = LinearRegression()
lr_selected.fit(X_train_scaled_selected, y_train)
mae_selected = mean_absolute_error(y_test, lr_selected.predict(X_test_scaled_selected))

print(f"\nMAE на всех признаках: {mae_all:.4f}")
print(f"MAE на признаках, выбранных DecisionTreeRegressor: {mae_selected:.4f}")
````
